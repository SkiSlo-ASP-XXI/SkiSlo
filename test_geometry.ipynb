```python
import laspy
import open3d as o3d
import numpy as np

filename = '/Users/andre/Documents/github.nosync/SkiSlo/data/surfaces/Sestriere_fotogrammetria_95000.las'
```

```python
# 1. Carica il file LAS o LAZ
las = laspy.read(filename)
```

```python

# 2. Estrai le coordinate x, y, z
# 3. Crea un oggetto PointCloud per Open3D
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(np.vstack((las.x, las.y, las.z)).transpose())
```

```python
# 4. (Opzionale) Se il file ha colori RGB, caricali
if hasattr(las, 'red'):
    # In LAS i colori sono spesso a 16-bit, li scaliamo a 8-bit/0-1 per Open3D
    pcd.colors = o3d.utility.Vector3dVector(np.vstack((las.red, las.green, las.blue)).transpose() / 65535.0)
```

```python

# 5. Visualizza la nuvola di punti
print("Visualizzazione in corso...")
o3d.visualization.draw_geometries([pcd])
```
